```python
from langchain_core.messages.human import HumanMessage
from dotenv import load_dotenv
load_dotenv()

from langchain.tools import tool
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from typing import Any
import os
import base64
```

# 得到图片url
要么使用网址，要么将本地图片构建为base64,构建地址

```python
with open("./img/sample.jpg", "rb") as image_file:
    img_bytes = image_file.read()

base64_str = base64.b64encode(img_bytes).decode("utf-8")

url = f"data:image/jpeg;base64,{base64_str}"
#url = "https://images.agriharvest.tw/wp-content/uploads/2025/03/%E7%8E%AB%E7%91%B025.jpg"
```

# 创建模型
qwen3.5-plus多模态模型

```python
model2 = init_chat_model(
    model="qwen3.5-plus",
    model_provider="openai",
    base_url="https://ws-5v4gpo6z9o3ebst9.cn-beijing.maas.aliyuncs.com/compatible-mode/v1",
    api_key=os.getenv("DASHSCOPE_API_KEY")
)
agent = create_agent(
    model2,
)
```

# 创建消息

```python
message = HumanMessage([
    {"type":"text","text":"这张图片中是什么花"},
    {"type":"image","url":url}
])
```

# 发起请求

```python
response: dict[str, Any] | Any = agent.stream(
    {"messages":[message]},
    stream_mode="messages"
)
for res,meta in response:
    if res.content:
        print(res.content,end="",flush=True)
```
