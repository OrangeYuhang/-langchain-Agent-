```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents.middleware import SummarizationMiddleware

middleware=SummarizationMiddleware(
    model="deepseek-chat",
    trigger=("messages",3),#("token",int),(fraction,float)
    keep=("messages",1)     #("token",int),(fraction,float)
)
```

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from langgraph.checkpoint.memory import InMemorySaver
from langchain_core.runnables import RunnableConfig

agent =create_agent(
    model="deepseek-chat",
    checkpointer=InMemorySaver(),
    middleware=[middleware]
    
)
config:RunnableConfig = {"configurable":{"thread_id":"thread_1"}}

agent.invoke({"messages":[HumanMessage("你好,我是meok😀")]},config)
agent.invoke({"messages":[HumanMessage("我喜欢的动物是猫猫🐱")]},config)
agent.invoke({"messages":[HumanMessage("我喜欢的运动是乒乓球🏓")]},config)

res = agent.invoke({"messages":[HumanMessage("你还记得我吗❓")]},config)

print(res)
```
